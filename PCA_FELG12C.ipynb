```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# ==================================================
# USER SETTINGS
# ==================================================

pc1_file = "5pca1.xvg"
pc2_file = "5pca2.xvg"

temperature = 300.0      # Kelvin
bins = 140               # FEL resolution
smooth_sigma = 1.0       # Gaussian smoothing

# ==================================================
# READ XVG FILES
# ==================================================

def read_xvg_col(fname):

    df = pd.read_csv(
        fname,
        sep=r"\s+",
        comment='@',
        header=None,
        engine='python'
    )

    df = df[~df[0].astype(str).str.startswith('#')]

    return df.iloc[:,1].astype(float).values


# Read PC1 and PC2
pc1 = read_xvg_col(pc1_file)
pc2 = read_xvg_col(pc2_file)

# Convert nm → Å
pc1 = pc1 * 10
pc2 = pc2 * 10

# ==================================================
# CALCULATE FREE ENERGY LANDSCAPE
# ==================================================

H, xedges, yedges = np.histogram2d(
    pc1,
    pc2,
    bins=bins
)

# Probability distribution
prob = H / np.sum(H)

# RT (kJ/mol)
R = 0.008314462618
RT = R * temperature

# Free Energy
F = -RT * np.log(prob + 1e-12)

# Smooth FEL
F = gaussian_filter(F, sigma=smooth_sigma)

# Set global minimum = 0
F = F - np.min(F)

# ==================================================
# PLOT SETTINGS
# ==================================================

label_font = {
    'family': 'Times New Roman',
    'weight': 'bold',
    'size': 24
}

tick_fontsize = 18

# Bin centers
xcenters = 0.5 * (xedges[:-1] + xedges[1:])
ycenters = 0.5 * (yedges[:-1] + yedges[1:])

extent = [
    xcenters.min(),
    xcenters.max(),
    ycenters.min(),
    ycenters.max()
]

# ==================================================
# PLOT FEL
# ==================================================

fig, ax = plt.subplots(figsize=(8,7))

# FEL heatmap
im = ax.imshow(
    F.T,
    origin='lower',
    extent=extent,
    cmap='viridis',
    aspect='auto'
)

# Optional contour lines
ax.contour(
    xcenters,
    ycenters,
    F.T,
    levels=15,
    colors='k',
    linewidths=0.3,
    alpha=0.3
)

# Labels
ax.set_xlabel(
    "PC1 (Å)",
    fontdict=label_font
)

ax.set_ylabel(
    "PC2 (Å)",
    fontdict=label_font
)

ax.set_title(
    "Complex_5_FEL",
    fontdict=label_font
)

# Tick formatting
ax.tick_params(
    axis='both',
    labelsize=tick_fontsize
)

for label in ax.get_xticklabels():
    label.set_fontname('Times New Roman')
    label.set_fontweight('bold')

for label in ax.get_yticklabels():
    label.set_fontname('Times New Roman')
    label.set_fontweight('bold')

# Colorbar
cbar = plt.colorbar(
    im,
    orientation='horizontal',
    pad=0.12
)

cbar.set_label(
    "Free Energy (kJ/mol)",
    fontdict=label_font
)

for label in cbar.ax.get_xticklabels():
    label.set_fontname('Times New Roman')
    label.set_fontweight('bold')
    label.set_fontsize(14)

plt.tight_layout()

plt.savefig(
    "Complex_5_FEL.png",
    dpi=600,
    bbox_inches='tight'
)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load PC1
pc1 = pd.read_csv("FDApc1.xvg", sep=r"\s+", comment='@', header=None)
pc1 = pc1[~pc1[0].astype(str).str.startswith('#')]
pc1_vals = pd.to_numeric(pc1[1], errors='coerce')

# Load PC2
pc2 = pd.read_csv("FDApc2.xvg", sep=r"\s+", comment='@', header=None)
pc2 = pc2[~pc2[0].astype(str).str.startswith('#')]
pc2_vals = pd.to_numeric(pc2[1], errors='coerce')

# Remove NaNs
mask = ~(pc1_vals.isna() | pc2_vals.isna())
pc1_vals = pc1_vals[mask]
pc2_vals = pc2_vals[mask]

# Parameters
T = 300.0
kB = 0.008314  # kJ mol^-1 K^-1

# 2D histogram
H, xedges, yedges = np.histogram2d(
    pc1_vals,
    pc2_vals,
    bins=100,
    density=True
)

# Avoid log(0)
H[H == 0] = np.nan

# Free energy
F = -kB * T * np.log(H)

# Shift minimum to zero
F = F - np.nanmin(F)

# Grid centers
xcenters = (xedges[:-1] + xedges[1:]) / 2
ycenters = (yedges[:-1] + yedges[1:]) / 2
X, Y = np.meshgrid(xcenters, ycenters)

# Plot FEL
plt.figure(figsize=(8,6))

contour = plt.contourf(
    X, Y, F.T,
    levels=30,
    cmap='jet'
)

cbar = plt.colorbar(contour)
cbar.set_label("Free Energy (kJ/mol)", fontsize=18)

# Labels
plt.xlabel(
    "PC1 (Å)",
    fontsize=22,
    fontweight='bold',
    fontname='Times New Roman'
)

plt.ylabel(
    "PC2 (Å)",
    fontsize=22,
    fontweight='bold',
    fontname='Times New Roman'
)

plt.title(
    "Complex_FDA_FEL",
    fontsize=24,
    fontweight='bold',
    fontname='Times New Roman'
)

plt.xticks(
    fontsize=18,
    fontweight='bold',
    fontname='Times New Roman'
)

plt.yticks(
    fontsize=18,
    fontweight='bold',
    fontname='Times New Roman'
)



plt.tight_layout()
plt.savefig("FDA_FEL.png", dpi=300)
plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.ndimage import gaussian_filter, minimum_filter

# Load PC1
pc1 = pd.read_csv("62pc1.xvg", sep=r"\s+", comment='@', header=None)
pc1 = pc1[~pc1[0].astype(str).str.startswith('#')]
pc1_vals = pc1[1].astype(float).values

# Load PC2
pc2 = pd.read_csv("62pc2.xvg", sep=r"\s+", comment='@', header=None)
pc2 = pc2[~pc2[0].astype(str).str.startswith('#')]
pc2_vals = pc2[1].astype(float).values

# Make sure lengths match
n = min(len(pc1_vals), len(pc2_vals))
pc1_vals = pc1_vals[:n]
pc2_vals = pc2_vals[:n]

# Build histogram
H, xedges, yedges = np.histogram2d(
    pc1_vals,
    pc2_vals,
    bins=100
)

# Smooth histogram
H = gaussian_filter(H, sigma=2)

# Avoid log(0)
H += 1e-12

# Free energy
kB = 0.008314462618
T = 300.0

P = H / np.sum(H)
F = -kB * T * np.log(P)
F -= np.min(F)

# Find local minima
local_min = (F == minimum_filter(F, size=3))
minima = np.argwhere(local_min)

# Grid centers
xcenters = (xedges[:-1] + xedges[1:]) / 2
ycenters = (yedges[:-1] + yedges[1:]) / 2

results = []

for i, j in minima:
    x = xcenters[i]
    y = ycenters[j]
    energy = F[i, j]

    # Closest trajectory frame
    dist = np.sqrt((pc1_vals - x)**2 + (pc2_vals - y)**2)
    frame = np.argmin(dist)

    results.append((energy, frame, x, y))

# Sort by energy
results.sort(key=lambda r: r[0])

print("Top 5 representative FEL basins:\n")

for idx, (energy, frame, x, y) in enumerate(results[:5], start=1):
    print(
        f"State {idx}: "
        f"Frame = {frame}, "
        f"PC1 = {x:.2f}, "
        f"PC2 = {y:.2f}, "
        f"Free Energy = {energy:.2f} kJ/mol"
    )

```
